```python
from crewai import Agent, Task, Crew, Process
from langchain_community.llms import Ollama

# Define LLM using Ollama
llm = Ollama(model="ministral-3:3b")
```

```python

# Create Agent
researcher = Agent(
    role="Research Assistant",
    goal="Say Hi",
    backstory="You are a welcoming Ai Agent",
    llm=llm,
    verbose=True
)

# Create Task
task = Task(
    description="Say Hello to the user.",
    expected_output="A polite greeting message.",
    agent=researcher
)

# Create Crew
crew = Crew(
    agents=[researcher],
    tasks=[task],
    verbose=True
)

# Run Crew
result = crew.kickoff()
print(result)
```

```python
researcher = Agent(
    role="Researcher",
    goal="Find accurate information about the given topic",
    backstory="You are a senior researcher who is very good at extracting concise insights.",
    llm=llm,
    verbose=True
)
```

```python
research_task = Task(
    description="Research the correct steps to install kubernetes on Ubuntu Virtual Machine.",
    expected_output="A bullet-point list of key benefits.",
    agent=researcher
)
```

```python
crew = Crew(
    agents=[researcher],
    tasks=[research_task],
    process=Process.sequential
)
```

```python
# Run the crew
result = crew.kickoff()
print(result)
```

```python
researcher = Agent(
    role="Helper",
    goal="Answer clearly and briefly",
    backstory="You are a senior competitive Programmer having knowledge in C++ with Data Structure and Algorithms?",
    llm=llm,
    verbose=True
)
research_task = Task(
    description="Write the code of Preorder Traversal of a Binary Tree in C++.",
    expected_output="Give me the code properly in C++, with a formatted code.",
    agent=researcher
)
crew = Crew(
    agents=[researcher],
    tasks=[research_task],
    process=Process.sequential
)
# Run the crew
result = crew.kickoff()
print(result)
```
